# Get To Know A Dataset: Neurological Disease Multimodal Data Registry (NDMDR)

The Neurological Disease Multimodal Data Registry (NDMDR) is a Stanford University initiative designed to support open, reproducible research in neurological and neuromuscular disorders through access to de-identified multimodal research datasets.

## About NDMDR

NDMDR brings together multimodal datasets that may include electroencephalography (EEG), magnetic resonance imaging (MRI), clinical measures, neurocognitive assessments, electrophysiology, and other disease-relevant data.

The registry is designed to support biomarker discovery, artificial intelligence and machine learning research, reproducibility, and cross-dataset analysis.

## Current datasets

Examples of datasets available through NDMDR include:

- **EEGConnect** — EEG data from individuals with myotonic dystrophy type 1.
- **PedsDMConnect** — multisite pediatric myotonic dystrophy type 1 data including neuroimaging, neurocognitive, clinical, and electrophysiological measures.

Additional neurological and neuromuscular datasets will be added as the registry expands.

## Dataset organization

NDMDR is organized by dataset and modality. Each dataset includes its own documentation describing available files, metadata, data-use requirements, and recommended analysis procedures.

Documentation: https://ndmdr.stanford.edu/documentation.html
